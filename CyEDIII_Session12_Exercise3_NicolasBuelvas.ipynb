### Problem Statement
Given the Context-Free Grammar $G = (V, \Sigma, R, S)$:

$$
G : \begin{cases}
S \to Ba \mid A \mid \lambda \\
A \to Aa \mid a \\
B \to bB \mid S
\end{cases}
$$

- **Variables ($V$):** $\{S, A, B\}$
- **Alphabet / Terminals ($\Sigma$, *Sigma*):** $\{a, b\}$
- **Empty string ($\lambda$, *Lambda*):** null transition
- **Start Symbol ($S$):** $S$

**Objective:** Construct an equivalent grammar $G'$ (*G-prime*) such that $L(G') = L(G)$ with all unit productions eliminated.

### Theoretical Approach
To eliminate unit productions from grammar $G$, we apply the standard four-phase transformation:
1. **Identify** all productions of the form $A \to B$ where $A, B \in V$ (*A, B in V*).
2. **Compute** the unit closure $\text{Unit}(X)$ for each variable $X \in V$ through reflexive and transitive derivations ($X \Rightarrow^* Y$, *X derives Y*).
3. **Extract** the original non-unit productions $\text{NonUnit}(X)$ for each variable.
4. **Construct** the new production set $P'$ (*P-prime*) where each variable inherits all non-unit productions from every symbol in its unit closure ($P'(X) = \bigcup_{Y \in \text{Unit}(X)} \text{NonUnit}(Y)$, *union over the closure*), directly bypassing intermediate unit transitions.

### Step-by-Step Derivation

#### **First: Identify Unit Productions**
A unit production is a rule transitioning directly between two variables ($A \to B, A, B \in V$):
- **$S \to A$** (*S produces A* — Unit)
- **$B \to S$** (*B produces S* — Unit)

*Note:* All remaining rules ($Ba, \lambda$ [*Lambda*], $Aa, a, bB$) contain terminals or empty word $\lambda$, hence they are non-unit.

---

#### **Second: Compute Unit Closures**
The unit closure $\text{Unit}(X)$ contains all variables reachable from $X$ via unit transitions ($X \Rightarrow^* Y$, *X derives Y*):
- **$\text{Unit}(A) = \{ A \}$** (no outgoing unit rules)
- **$\text{Unit}(S) = \{ S, A \}$** (via $S \to A$)
- **$\text{Unit}(B) = \{ B, S, A \}$** (via chain $B \Rightarrow S \Rightarrow A$, *B derives S derives A*)

| Variable $X$ | Derivation Path ($\Rightarrow^*$, *derives*) | $\text{Unit}(X)$ |
| :--- | :--- | :--- |
| **$A$** | $A$ | $\{ A \}$ |
| **$S$** | $S \Rightarrow A$ | $\{ S, A \}$ |
| **$B$** | $B \Rightarrow S \Rightarrow A$ | $\{ B, S, A \}$ |

---

#### **Third: Identify Original Non-Unit Productions**
We extract all productions with terminals or $\lambda$ (*Lambda*):
- **$\text{NonUnit}(S) = \{ Ba, \lambda \}$**
- **$\text{NonUnit}(A) = \{ Aa, a \}$**
- **$\text{NonUnit}(B) = \{ bB \}$**

---

#### **Finally: Construct Equivalent Grammar $G'$ (*G-prime*)**
Each variable $X$ inherits the non-unit productions from all $Y \in \text{Unit}(X)$:

1. **For $S$** ($\text{Unit}(S) = \{S, A\}$):
   - From $S$: $Ba, \lambda$
   - From $A$: $Aa, a$
   $$\mathbf{S \to Ba \mid Aa \mid a \mid \lambda}$$

2. **For $A$** ($\text{Unit}(A) = \{A\}$):
   - From $A$: $Aa, a$
   $$\mathbf{A \to Aa \mid a}$$

3. **For $B$** ($\text{Unit}(B) = \{B, S, A\}$):
   - From $B$: $bB$
   - From $S$: $Ba, \lambda$
   - From $A$: $Aa, a$
   $$\mathbf{B \to bB \mid Ba \mid Aa \mid a \mid \lambda}$$

---

### Resulting Grammar $G'$ (*G-prime*)
$$
G' : \begin{cases}
S \to Ba \mid Aa \mid a \mid \lambda \\
A \to Aa \mid a \\
B \to bB \mid Ba \mid Aa \mid a \mid \lambda
\end{cases}
$$

```python
# Verification using pyformlang
!pip install -q pyformlang

from pyformlang.cfg import CFG

# Original Grammar G
cfg = CFG.from_text("""
S -> B a
S -> A
S -> 
A -> A a
A -> a
B -> b B
B -> S
""")

print("--- Original Productions ---")
for p in sorted(cfg.productions, key=lambda x: str(x.head)):
    print(p)

# Algorithmic elimination of unit productions
cfg_no_unit = cfg.eliminate_unit_productions()

print("\n--- Grammar G-prime (Unit Productions Eliminated) ---")
for p in sorted(cfg_no_unit.productions, key=lambda x: str(x.head)):
    print(p)
```
