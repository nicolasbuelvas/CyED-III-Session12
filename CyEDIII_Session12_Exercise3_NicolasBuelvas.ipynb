### Problem Statement
Context-Free Grammar G:

$$
G : \begin{cases}
S \to Ba \mid A \mid \lambda \\
A \to Aa \mid a \\
B \to bB \mid S
\end{cases}
$$

- Variables (V): {S, A, B}
- Alphabet ($\Sigma$, Sigma): {a, b}
- Empty string ($\lambda$, Lambda)
- Start symbol: S

Goal: Eliminate all unit productions to obtain an equivalent grammar $G'$ (G-prime) such that $L(G') = L(G)$.

### Theoretical Approach
Standard four-step procedure:
1. Identify all unit productions ($A \to B$, with $A, B \in V$).
2. Compute unit closures $\text{Unit}(X)$ through unit derivations ($X \Rightarrow^* Y$).
3. List original non-unit productions $\text{NonUnit}(X)$.
4. Construct new production rules $P'$ (P-prime) by inheritance ($P'(X) = \bigcup_{Y \in \text{Unit}(X)} \text{NonUnit}(Y)$).

### Step-by-Step Derivation

#### First: Identify Unit Productions
A unit production has the form $A \to B$ ($A, B \in V$):
- $S \to A$
- $B \to S$

All other rules contain terminals from $\Sigma$ (Sigma) or $\lambda$ (Lambda), so they are non-unit.

#### Second: Compute Unit Closures
$\text{Unit}(X)$ contains all variables reachable through unit transitions ($X \Rightarrow^* Y$):
- $\text{Unit}(A) = \{ A \}$
- $\text{Unit}(S) = \{ S, A \}$
- $\text{Unit}(B) = \{ B, S, A \}$ (via chain $B \Rightarrow S \Rightarrow A$)

| Variable $X$ | Derivation Path ($\Rightarrow^*$) | $\text{Unit}(X)$ |
| :--- | :--- | :--- |
| $A$ | $A$ | $\{ A \}$ |
| $S$ | $S \Rightarrow A$ | $\{ S, A \}$ |
| $B$ | $B \Rightarrow S \Rightarrow A$ | $\{ B, S, A \}$ |

#### Third: Identify Original Non-Unit Productions
Extract original non-unit terminal rules:
- $\text{NonUnit}(S) = \{ Ba, \lambda \}$
- $\text{NonUnit}(A) = \{ Aa, a \}$
- $\text{NonUnit}(B) = \{ bB \}$

#### Finally: Construct Equivalent Grammar G'
Each variable inherits non-unit productions from all variables in its unit closure:

1. For S (from S and A):
$$\mathbf{S \to Ba \mid Aa \mid a \mid \lambda}$$

2. For A (from A):
$$\mathbf{A \to Aa \mid a}$$

3. For B (from B, S, and A):
$$\mathbf{B \to bB \mid Ba \mid Aa \mid a \mid \lambda}$$

### Resulting Grammar G' (G-prime)

$$
G' : \begin{cases}
S \to Ba \mid Aa \mid a \mid \lambda \\
A \to Aa \mid a \\
B \to bB \mid Ba \mid Aa \mid a \mid \lambda
\end{cases}
$$

```python
# Verification using pyformlang
!pip install -q pyformlang

from pyformlang.cfg import CFG

cfg = CFG.from_text("""
S -> B a
S -> A
S -> 
A -> A a
A -> a
B -> b B
B -> S
""")

print("--- Original Grammar G ---")
for p in sorted(cfg.productions, key=lambda x: str(x.head)):
    print(p)

cfg_no_unit = cfg.eliminate_unit_productions()

print("\n--- Resulting Grammar G-prime ---")
for p in sorted(cfg_no_unit.productions, key=lambda x: str(x.head)):
    print(p)
```
