### Problem Statement
Grammar G:

$$
G : \begin{cases}
S \to Ba \mid A \mid \lambda \\
A \to Aa \mid a \\
B \to bB \mid S
\end{cases}
$$

Variables (V): {S, A, B}
Alphabet ($\Sigma$, Sigma): {a, b}
Empty string ($\lambda$, Lambda)
Start symbol: S

Goal: Remove all unit productions to find an equivalent grammar $G'$ (G-prime).

### Step 1: Identify Unit Productions
A unit production has the form $A \to B$, with $A, B \in V$.

- S -> A
- B -> S

All other rules contain terminals or $\lambda$, so they are non-unit.

### Step 2: Compute Unit Closures
$\text{Unit}(X)$ contains all variables reachable from $X$ using unit transitions ($X \Rightarrow^* Y$):

- Unit(A) = {A}
- Unit(S) = {S, A}
- Unit(B) = {B, S, A} via chain B => S => A

### Step 3: Original Non-Unit Productions
Collect non-unit productions for each variable:

- NonUnit(S) = {Ba, $\lambda$}
- NonUnit(A) = {Aa, a}
- NonUnit(B) = {bB}

### Step 4: Construct Grammar G' (G-prime)
Each variable inherits the non-unit productions of all symbols in its unit closure:

For S (from S and A):
S -> Ba | Aa | a | $\lambda$

For A (from A):
A -> Aa | a

For B (from B, S, and A):
B -> bB | Ba | Aa | a | $\lambda$

### Resulting Grammar G'

$$
G' : \begin{cases}
S \to Ba \mid Aa \mid a \mid \lambda \\
A \to Aa \mid a \\
B \to bB \mid Ba \mid Aa \mid a \mid \lambda
\end{cases}
$$
